**Analytical grain: one row per client-account relationship.**  
The business question concerns customer engagement, so `client_id` identifies the customer of interest. However, most behavioural activity is recorded at `account_id` level. The `disp` table therefore provides the bridge between customer identity and account-level behaviour.

**Lifecycle start: account.date**  
Account creation represents the beginning of the customer-account relationship and provides a consistent acquisition point from which subsequent product behaviour can be measured.

**Observation window: first 60 days after account creation.**  
Behaviour occurring during this period will be used to characterize each client-account relationship and later construct predictive features.

**Behavioural Signal Mapping**  
Behaviour during the first 60 days after account creation will be used to characterize early product adoption. Candidate behavioural domains are assessed for both business relevance and temporal validity before being included in the predictive feature set.

| Domain | Source | Potential Signal | Status |
|---|---|---|---|
| Transactions | `trans.csv` | Account activity, transaction frequency, monetary flows and types of financial activity | Included |
| Balance | `trans.csv` | Level and evolution of funds held in the account | Included / to engineer |
| Standing orders | `order.csv` | Adoption of recurring payments and integration of Vault into regular financial activity | Excluded |
| Cards | `card.csv` | Card adoption and timing of card issuance | Excluded |
| Loans | `loan.csv` | Adoption of credit products and broader product usage | To assess |

Transactions and balance are treated as the primary behavioural signals because they most directly represent ongoing account usage and can be evaluated within the defined observation window. Other product-adoption sources are retained only when their timing can be established relative to account creation.

**Feature Source Limitations**

**Standing Orders**  
Standing orders were initially considered as a potential signal of early product adoption. However, `order.csv` does not contain a creation or activation date for each order.

As a result, it is not possible to determine whether a standing order existed within the first 60 days after account creation. Using these records as early-behaviour predictors could therefore introduce temporal leakage.

Standing-order features are excluded from the 60-day predictive feature set.

**Card Adoption**  
Card issuance was initially considered as a potential early product-adoption signal. Analysis of the time between account creation and card issuance showed that the earliest observed card was issued **203 days after account creation**.

Therefore, no card issuance occurs within the defined 60-day observation window. Card-related features would provide no variation during the observation period, while using later issuance information would introduce temporal leakage.

Card-related features are therefore excluded from the 60-day predictive feature set.

**Analytical Limitation: Shared Accounts**  
The same `account_id` may be associated with two dispositions: an `OWNER` and a `DISPONENT`, representing two different clients with access to the same account.

Most behavioural data, including transactions and balances, is recorded at the `account_id` level. Therefore, for shared accounts, it is not possible to determine which individual client generated a specific behaviour.

Account-level behaviour will consequently be attributed to the client-account relationship, and results involving shared accounts should not be interpreted as individual-level behavioural attribution.
